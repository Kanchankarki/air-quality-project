```python
def pm25_to_aqi(pm25):
    if pm25 <= 30:
        Clow, Chigh, Ilow, Ihigh = 0, 30, 0, 50
    elif pm25 <= 60:
        Clow, Chigh, Ilow, Ihigh = 31, 60, 51, 100
    elif pm25 <= 90:
        Clow, Chigh, Ilow, Ihigh = 61, 90, 101, 200
    elif pm25 <= 120:
        Clow, Chigh, Ilow, Ihigh = 91, 120, 201, 300
    elif pm25 <= 250:
        Clow, Chigh, Ilow, Ihigh = 121, 250, 301, 400
    else:
        Clow, Chigh, Ilow, Ihigh = 251, 500, 401, 500

    aqi = ((Ihigh - Ilow) / (Chigh - Clow)) * (pm25 - Clow) + Ilow
    return round(aqi)
```

```python
data_hourly['AQI']=data_hourly['pm25'].apply(pm25_to_aqi)
```

```python
features=data_hourly[['o3','pm25']].values
```

```python
features
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler(feature_range=(0,1))
scaled_features=scaler.fit_transform(features)
```

```python
#prepare seq for lstm
def seq(data,time_steps=24):
    X,Y=[],[]
    for i in range(time_steps,len(data)):
        X.append(data[i-time_steps:i])
        Y.append(data[i,1])
    return np.array(X) ,np.array(Y)

time_steps=24
X,Y=seq(scaled_features,time_steps)
```

```python
X.shape
```

```python
Y.shape
```

```python
#split
split=int(0.8 *len(X))
X_train,X_test = X[:split],X[split:]
Y_train,Y_test = Y[:split],Y[split:]
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM,Dense,Dropout
```

```python
num_feat=X_train.shape[2]
```

```python
num_feat
```

```python
ts=X_train.shape[1]
ts
```

```python
#model architecture
model = Sequential([
    #1sr
    LSTM(64,activation='tanh',
    return_sequences=True,
    input_shape=(time_steps, num_feat)),
    Dropout(0.2),

    #2nd
    LSTM(32,activation='tanh'),
    Dropout(0.2),

    #Final
    Dense(1,activation='linear')
])
```

```python
#model compile
model.compile(optimizer='adam',loss='mean_squared_error')
```

```python
model.summary()
```

```python
#train
result=model.fit(X_train,Y_train,
                 epochs=50,batch_size=32,
                 validation_data=(X_test,Y_test),
                 verbose=1)
```

```python
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
y_test_scaled=np.zeros((len(Y_test),num_feat))
y_test_scaled[:,1]=Y_test
y_pred_scaled=np.zeros((len(y_pred),num_feat))
y_pred_scaled[:,1]=y_pred[:,0]

y_test_actual=scaler.inverse_transform(y_test_scaled)[:,1]
y_pred_actual=scaler.inverse_transform(y_pred_scaled)[:,1]
```

```python
plt.figure(figsize=(15, 6))
plt.plot(y_test_actual, label='Actual', color='blue') # Show 1 week (168 hours)
plt.plot(y_pred_actual, label='LSTM Predicted', color='red', linestyle='--')
plt.title('Kathmandu Air Quality')
plt.xlabel('Hours')
plt.ylabel('AQI')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
mae = mean_absolute_error(y_test_actual, y_pred_actual)
rmse = np.sqrt(mean_squared_error(y_test_actual, y_pred_actual))
r2 = r2_score(y_test_actual, y_pred_actual)
```

```python
scaler1 = MinMaxScaler()
scaled_features1 = scaler1.fit_transform(features)
```

```python
def create_sequences(data, time_steps=24):
    X, y = [], []
    for i in range(time_steps, len(data)):
        X.append(data[i-time_steps:i])
        y.append(data[i, 1])  # PM25_log is index 1
    return np.array(X), np.array(y)

X, y = create_sequences(scaled_features1, time_steps=24)

print(X.shape, y.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(24, 2)),
    Dropout(0.2),
    LSTM(32),
    Dropout(0.2),
    Dense(1)  # log(PM25)
])

model.compile(optimizer="adam", loss="mse")
model.fit(X_train, y_train, epochs=40, batch_size=32, validation_split=0.1)

model.save("aqi_lstm_log_model.h5")
```
